```python
sensor_cols = [
    col for col in df.columns
    if col.startswith("sensor_")
]

print(sensor_cols)
```

```python
variance = (
    df[sensor_cols]
    .var()
    .sort_values()
)

variance
```

```python
plt.figure(figsize=(12,5))

variance.plot(kind="bar")

plt.title("Sensor Variance")
plt.show()
```

```python
corr_rul = (
    df[sensor_cols]
    .corrwith(df["RUL"])
    .sort_values()
)

corr_rul
```

```python
plt.figure(figsize=(8,8))

corr_rul.plot(kind="barh")

plt.title("Sensor Correlation with RUL")
plt.show()
```

```python
engine = df[df["engine_id"] == 1]
important = [
    "sensor_11",
    "sensor_4",
    "sensor_13",
    "sensor_8"
]

for sensor in important:
    plt.figure(figsize=(10,4))

    plt.plot(
        engine["cycle"],
        engine[sensor]
    )

    plt.title(sensor)

    plt.xlabel("Cycle")
    plt.ylabel("Value")

    plt.show()
```

```python
sns.scatterplot(
    data=df.sample(5000),
    x="RUL",
    y="sensor_11"
)

plt.title("Sensor 11 vs RUL")
plt.show()
```

```python
corr_matrix = df[sensor_cols].corr()

plt.figure(figsize=(14,10))

sns.heatmap(
    corr_matrix,
    cmap="coolwarm"
)

plt.title("Sensor Correlation Matrix")

plt.show()
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

X = df[sensor_cols]

X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

plot_df = pd.DataFrame({
    "PC1": X_pca[:,0],
    "PC2": X_pca[:,1],
    "health_stage": df["health_stage"]
})
```

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    data=plot_df,
    x="PC1",
    y="PC2",
    hue="health_stage",
    alpha=0.5
)

plt.show()
```

```python
sns.scatterplot(
    data=plot_df,
    x="PC1",
    y="PC2",
    hue=df["RUL"],
    palette="viridis"
)
```

```python
pca.explained_variance_ratio_
```

```python
sum(pca.explained_variance_ratio_)
```

```python
loadings = pd.DataFrame(
    pca.components_.T,
    columns=["PC1", "PC2"],
    index=sensor_cols
)

loadings
```
